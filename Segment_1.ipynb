### Bank Robber Example
- The sheriff is moving 180 km/h
- The bank robber is moving 150 km/h and has a five-minute head start
- Calculate how long it takes the shreiff to catch the robber and how far they have traveled when the robber is captured
- Display all of this on a graph

````python

# Time simulation:
time = np.linspace(0, 40, 1000) # Start, End, n points

distance_robber = 2.5 * time
distance_sheriff = 3 * (time-5)


# Plotting a graph
figure, axes = plt.subplots()

plt.title("A Bank Robber Caught:")
plt.xlabel("time (in minutes)")
plt.ylabel("distance (in km)")

axes.set_xlim([0, 40])
axes.set_ylim([0, 100])

axes.plot(time, distance_robber, c="green")
axes.plot(time, distance_sheriff, c="red")

plt.axvline(x = 30, c = "purple", linestyle = "--")
plt.axhline(y = 75, c = "purple", linestyle = "--")

plt.show()
````

### Solar Panels Exercise
- On April 1st, the Mark I solar panel generates 1 kJ/day.
- On May 1st, the Mark II solar panel generates 4 kJ/day.
- Calculate the amount of days it takes mark II to generate the same amount of power mark I has generated.
- Calculate how much energy has been generated
- Display all of this on a graph

````python
days = np.linspace(0, 60 , 1000)

energy_m1 = 1 * days
energy_m2 = 4 * (days - 30)

figure, axes = plt.subplots()

plt.title("Solar Panles Energy Generation:")
plt.xlabel("time (in days)")
plt.ylabel("energy (in kJ)")

axes.plot(days, energy_m1, label = "Mark I", c = "green")
axes.plot(days, energy_m2, label = "Mark II", c = "red")

axes.set_xlim([0, 60])
axes.set_ylim([0, 100])

plt.axvline(x = 40, c = "purple", linestyle = "--")
plt.axhline(y = 40, c = "purple", linestyle = "--")

plt.show()
````

### Vector Transposition

````python
import numpy as np

# Linear Vector
linear_vector = np.array([25, 2, 5]) # 1 dimensional

linear_vector_transposition = linear_vector.T

linear_vector_transposition
linear_vector_transposition.shape  # This won't show any change, since this is a lineaer array, meaning there is no other dimension fr this to transpose into


# Matrix Vector
matrix_vector = np.array([[25, 2, 5]]) # 2 dimensional

matrix_vector_transposition = matrix_vector.T

matrix_vector_transposition
matrix_vector_transposition.shape 
````
